```python
import json
from pathlib import Path
import pandas as pd

# Paths
VIDEO_DIR = Path(r"C:\ego4d\v2\full_scale")
ANNOTATION_FILE = Path(r"C:\ego4d\v2\annotations\narration.json")

# Load narration JSON
with open(ANNOTATION_FILE, "r", encoding="utf-8") as f:
    data = json.load(f)

# Get downloaded video IDs from filenames
video_files = list(VIDEO_DIR.glob("*.mp4"))

downloaded_videos = {
    video.stem: video
    for video in video_files
}

print(f"Downloaded videos found: {len(downloaded_videos)}")

# Extract narration rows
rows = []

for video_uid, video_data in data.items():
    if video_uid in downloaded_videos:
        narrations = video_data.get("narration_pass_1", {}).get("narrations", [])

        for item in narrations:
            rows.append({
                "video_uid": video_uid,
                "video_path": str(downloaded_videos[video_uid]),
                "timestamp_sec": item.get("timestamp_sec"),
                "narration_text": item.get("narration_text")
            })

```

```python
df = pd.DataFrame(rows)

print(f"Matched narration rows: {len(df)}")
df.head()
```
